```python
import openai
import pandas as pd
import tkinter as tk
from tkinter import simpledialog, messagebox

# 🔑 Ta clé OpenAI
openai.api_key = "Votre clé API Open AI"

# 📁 Créons le fichier Excel si pas encore là
def creer_fichier_excel():
    data = {
        "Maladie": ["Grippe", "Paludisme", "Possession démoniaque"],
        "Symptôme principal": ["Fièvre, toux", "Fièvre, frissons", "Cris, voix étranges"],
        "Traitement": ["Paracétamol, repos", "Antipaludéens", "Exorcisme, prières"]
    }
    df = pd.DataFrame(data)
    df.to_excel("data.xlsx", index=False)

# 📄 Lire les données Excel et créer un prompt pour l'IA
def lire_excel_et_envoyer_question(question):
    df = pd.read_excel("data.xlsx")
    contenu = df.to_string(index=False)

    prompt = f"""
Tu es un assistant intelligent. Voici des données médicales :

{contenu}

Question utilisateur : {question}
Réponds de façon précise avec les infos du tableau.
    """

    try:
        response = openai.ChatCompletion.create(
            model="gpt-4",
            messages=[
                {"role": "system", "content": "Tu es un expert médical."},
                {"role": "user", "content": prompt}
            ],
            temperature=0.3
        )
        return response.choices[0].message["content"].strip()

    except Exception as e:
        return f"Erreur : {str(e)}"

# 💬 Interface graphique
def lancer_interface():
    creer_fichier_excel()

    root = tk.Tk()
    root.withdraw()
    question = simpledialog.askstring("Interrogation IA", "Pose ta question :")

    if question:
        reponse = lire_excel_et_envoyer_question(question)
        messagebox.showinfo("Réponse de l'IA", reponse)
    else:
        messagebox.showinfo("Info", "Aucune question posée.")

# ▶️ Lancer l'app
if __name__ == "__main__":
    lancer_interface()
```
